## Top 10 Cities

```python
plt.figure(figsize=(12,6))
df['City'].value_counts().head(10).plot(kind='bar')
plt.title('Top 10 Cities');plt.xticks(rotation=45);plt.show()
```

## Top 10 Cuisines

```python
top=(df['Cuisines'].dropna().str.split(',').explode().str.strip().value_counts().head(10))
plt.figure(figsize=(10,5))
sns.barplot(x=top.values,y=top.index)
plt.show()
```

## Rating Distribution

```python
sns.histplot(df['Aggregate rating'],bins=20,kde=True);plt.show()
```

## Online Delivery

```python
sns.countplot(data=df,x='Has Online delivery');plt.show()
```

## Table Booking

```python
sns.countplot(data=df,x='Has Table booking');plt.show()
```

## Price Range

```python
sns.countplot(data=df,x='Price range');plt.show()
```

## Correlation Heatmap

```python
num=df.select_dtypes(include=np.number)
sns.heatmap(num.corr(),annot=True,cmap='coolwarm')
plt.show()
```

## Restaurant Locations

```python
fig=px.scatter_mapbox(df,lat='Latitude',lon='Longitude',hover_name='Restaurant Name',hover_data=['Aggregate rating'],zoom=3,height=600)
fig.update_layout(mapbox_style='open-street-map')
fig.show()
```

## Average Rating by City

```python
df.groupby('City')['Aggregate rating'].mean().sort_values(ascending=False).head(10).plot(kind='bar')
plt.show()
```
